```python
import numpy as np
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
labels = ["A postive review " , " A negative review "]

samplereviews = ["A  movies is not good " , " Vfs was good and music was excellent" , " good a movie by director"]
```

```python
embed_reviews = model.encode(samplereviews)
embed_labels = model.encode(labels)
print(f"{embed_reviews.shape} and  {embed_labels.shape}")
print(embed_reviews[0][:10])  # lest print frist 10 vector values of first review
```

```python
# to under stand how normailization happen with sample matrix
sample_arr = np.array([
    [2.3 , 3.4 , 5.2 ] ,
    [1.3 , 2.5 , 2.6] ,
    [2.4 , 6.3 , 5.6]
])
print(sample_arr.shape)
print(sample_arr)

# norm of each in the arr

norms_each_row = np.linalg.norm(sample_arr , axis=1 , keepdims = True )
print(norms_each_row.shape)
print(norms_each_row)

normalized_vecotrs =  sample_arr / norms_each_row
print(normalized_vecotrs.shape)
print(normalized_vecotrs )


print(np.linalg.norm(normalized_vecotrs , axis= 1 , keepdims=True ))
```

```python
# now l2 normialzation

embed_reviews = embed_reviews /np.linalg.norm(embed_reviews , axis = 1 , keepdims = True)
embed_labels = embed_labels /np.linalg.norm(embed_labels , axis = 1 , keepdims = True)
print(embed_reviews.shape)
print(embed_labels.shape)
```

```python
scores = np.dot(embed_reviews , embed_labels.T)
print(scores.shape)
print(scores.argmax(axis=1))
for  ind in scores.argmax(axis=1) :
   print(labels[ind])
```

```python
# now creating same in function which takes the user revies return the postive review or negative review

def  check( review : str ) :
  review_vector = model.encode(review)
  review_vector = review_vector / np.linalg.norm(review_vector, keepdims=True)
  scores_each_label = np.dot(review_vector , embed_labels.T)
  ind = scores_each_label.argmax()
  print(scores_each_label)
  return labels[ind]
```

```python
print(check("this movie is good but the music is not good"))
```
